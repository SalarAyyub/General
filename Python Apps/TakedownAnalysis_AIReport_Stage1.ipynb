```python
#task 2: Descriptive Analysis of each column

data = pd.DataFrame(df)

def metric_analysis(column): 

    descriptive_analysis = data[column].describe()
    print(f"\nStats for '{column}': ")
    print(descriptive_analysis)

    sbn.histplot(data[column])
    
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.title(f"Histogram of {column}")
    plt.show()


for cols in data.columns:
    metric_analysis(cols)

    


    
    
```

```python
#task 3: Normal Distribution of each of the columns

def normal_distribution(column):

    norDist = data[column].describe()
    print(f"\nNormal Distribution for '{column}: '")
    print(norDist)

    plt.figure(figsize=(8,5))
    n, bins, patches = plt.hist(data[column], density = True, alpha = 0.05)

    mu = norDist['mean']
    sigma = norDist['std']
    x = np.linspace(bins.min(), bins.max(), 100)
    y = norm.pdf(x, mu, sigma)

    plt.plot(x,y, label = "Normal Distributions")
    plt.xlabel(column)
    plt.ylabel("Probability Density")
    plt.title(f"Normal Distribution for {column}")
    plt.legend()
    plt.grid(True)
    plt.show()

for cols in data.columns:
    normal_distribution(cols)

    
```
