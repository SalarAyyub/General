```python
#operation 1: descriptive analysis of numeric (excluding ID) columns

data = pd.DataFrame(dataset)

def metric_analysis(data, column): 

    if column not in ["fighter_id","name", "natinality_name"]:
       descriptive_analysis = data[column].describe()
       print(f"\nStats for '{column}': ")
       print(descriptive_analysis)

       sns.histplot(data[column])
    
       plt.xlabel(column)
       plt.ylabel("Frequency")
       plt.title(f"Histogram of {column}")
       plt.show()


for cols in data.columns:
    metric_analysis(data.copy(), cols)
```

```python
#normal distribution of the selected columns

def normal_distribution(data, column):

    if column not in ["fighter_id","name", "natinality_name"]:
       refined_data = data.fillna(0)
       norDist = refined_data[column].describe()
       print(f"\nNormal Distribution for '{column}: '")
       print(norDist)

       plt.figure(figsize=(8,5))
       n, bins, patches = plt.hist(refined_data[column], density = True, alpha = 0.05)

       mu = norDist['mean']
       sigma = norDist['std']
       x = np.linspace(bins.min(), bins.max(), 100)
       y = norm.pdf(x, mu, sigma)

       plt.plot(x,y, label = "Normal Distributions")
       plt.xlabel(column)
       plt.ylabel("Probability Density")
       plt.title(f"Normal Distribution for {column}")
       plt.legend()
       plt.grid(True)
       plt.show()

for cols in data.columns:
    normal_distribution(data.copy(), cols)
```

```python
#Defining Thresholds w.r.t 'overall_score'
#threshold = Substandard[0.0, 0.1], Vulnerable[0.1,0.25],... 
#Par[0.25,0.4], Standard[0.4,0.65], Strong[0.65,0.85], Top Tier[0.85, 1.0]

data['overall_score'] = data["overall_score"].fillna(0)

thresholds = [0.0, 0.1, 0.25, 0.4, 0.65, 0.85, 1.0]
labels = ['Substandard', 'Vunerable', 'Par', 'Standard', 'Strong', 'Top Tier']

def threshold_category_column(df, column): 
     
    if column in ['overall_score']:
        quartiles = df[column].describe(percentiles=thresholds)[thresholds]
        print(quartiles)
        
        if not quartiles.empty:

            adjusted_thresholds = quartiles.tolist() + [1.0]
            #adjusted_thresholds = [round(p+0.005,2) for p in quartiles.tolist()] + [0.1]
            iqr = adjusted_thresholds[3] - adjusted_thresholds[1]

            substandard = adjusted_thresholds[1] - 1.5*iqr
            vulnerable = adjusted_thresholds[1] + 0.5*iqr
            par = adjusted_thresholds[1] + 1.5*iqr
            standard = adjusted_thresholds[3] + 1.5*iqr
            top_tier = adjusted_thresholds[3] + 2.5*iqr
    
            def labeling(x):
                return pd.cut(x, adjusted_thresholds, labels=labels[:-1])
   
            df['threshold_category'] = labeling(df[column])
            
        else: 
            return -404
        
    
    return df



try:
   newDF = threshold_category_column(data.copy(), 'overall_score')
except Exception as e: 
    print(f"Error: {e}")

if newDF is not None:        
    print(newDF)
    
```
